```python
import cv2
import numpy as np

# ==========================================
# PARTE 1: Criando as imagens sintéticas do enigma
# (Apenas para podermos testar o código, já que não temos as originais)
# ==========================================
def criar_imagens_teste():
    # Cria uma tela branca de 200x200
    # A cor preta é (0,0,0) e a branca é (255,255,255)
    img1 = np.ones((200, 200), dtype=np.uint8) * 255
    img2 = np.ones((200, 200), dtype=np.uint8) * 255
    img3 = np.ones((200, 200), dtype=np.uint8) * 255

    # --- Figura 1: 3 quadrados na esquerda, 3 na direita (vamos desenhar 3 preenchidos) ---
    # Desenha 3 quadrados pretos na coluna da esquerda
    cv2.rectangle(img1, (20, 20), (80, 80), (0, 0, 0), -1)
    cv2.rectangle(img1, (20, 90), (80, 150), (0, 0, 0), -1)
    cv2.rectangle(img1, (20, 160), (80, 190), (0, 0, 0), -1)
    # Para simplificar o teste, vamos desenhar apenas os preenchidos.
    # O algoritmo do OpenCV vai ignorar o fundo branco e contar apenas os pretos.

    # --- Figura 2: 5 quadrados preenchidos (1 na esquerda, 4 na direita) ---
    cv2.rectangle(img2, (20, 20), (80, 190), (0, 0, 0), -1) # Quadrado grande na esquerda
    cv2.rectangle(img2, (100, 20), (170, 80), (0, 0, 0), -1) # Direita cima
    cv2.rectangle(img2, (100, 90), (170, 150), (0, 0, 0), -1) # Direita meio
    cv2.rectangle(img2, (100, 160), (170, 190), (0, 0, 0), -1) # Direita baixo

    # --- Figura 3: 7 quadrados preenchidos (1 em cima, 1 embaixo) ---
    # Como o desenho original tem 7 preenchidos, vamos simular 7 quadrados pretos
    cv2.rectangle(img3, (20, 20), (80, 80), (0, 0, 0), -1)
    cv2.rectangle(img3, (100, 20), (170, 80), (0, 0, 0), -1)
    cv2.rectangle(img3, (20, 90), (80, 150), (0, 0, 0), -1)
    cv2.rectangle(img3, (100, 90), (170, 150), (0, 0, 0), -1)
    cv2.rectangle(img3, (20, 160), (80, 190), (0, 0, 0), -1)
    cv2.rectangle(img3, (100, 160), (170, 190), (0, 0, 0), -1)
    cv2.rectangle(img3, (60, 80), (100, 120), (0, 0, 0), -1) # Quadrado central extra

    # Salva as imagens
    cv2.imwrite('figura1.png', img1)
    cv2.imwrite('figura2.png', img2)
    cv2.imwrite('figura3.png', img3)
    print("Imagens de teste criadas com sucesso!")

# ==========================================
# PARTE 2: O "Cérebro" do computador (Visão Computacional)
# ==========================================
def contar_quadrados_preenchidos(caminho_imagem):
    # 1. Carrega a imagem em escala de cinza
    img = cv2.imread(caminho_imagem, cv2.IMREAD_GRAYSCALE)

    # 2. Binarização (Thresholding)
    # Transforma a imagem em preto e branco puro.
    # Tudo que for escuro (os quadrados) vira 255 (branco) e o fundo vira 0 (preto).
    _, thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)

    # 3. Encontra os contornos (as bordas dos quadrados)
    contornos, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # 4. Filtro de área (para ignorar sujeirinhas ou ruídos)
    # Só conta se a área do contorno for maior que 100 pixels
    quadrados_validos = [c for c in contornos if cv2.contourArea(c) > 100]

    return len(quadrados_validos)

# ==========================================
# PARTE 3: A SUA REGRA DE PROGRESSÃO
# ==========================================
def sua_logica_de_progressao(qtd_preenchidos):
    # A base é a Figura 1: 3 preenchidos = 18
    # Regra: A cada +1 preenchido, o valor sobe +1
    valor_final = 18 + (qtd_preenchidos - 3)
    return valor_final

# ==========================================
# PARTE 4: Executando tudo
# ==========================================
criar_imagens_teste()

print("\n--- Resolvendo o Enigma com Visão Computacional ---")

for i in range(1, 4):
    arquivo = f'figura{i}.png'

    # O computador "lê" a imagem e conta
    qtd = contar_quadrados_preenchidos(arquivo)

    # A sua lógica calcula o valor
    resultado = sua_logica_de_progressao(qtd)

    print(f"Figura {i}: Encontrou {qtd} quadrados preenchidos -> Resultado: {resultado}")
```

```python
import cv2
import numpy as np

# ==========================================
# PARTE 1: Criando imagens com ESPAÇO entre os quadrados
# ==========================================
def criar_imagens_teste():
    img1 = np.ones((200, 200), dtype=np.uint8) * 255
    img2 = np.ones((200, 200), dtype=np.uint8) * 255
    img3 = np.ones((200, 200), dtype=np.uint8) * 255

    # Figura 1: 3 quadrados na esquerda, 3 na direita (com espaço entre eles)
    # Desenhamos os 3 preenchidos da esquerda
    for i in range(3):
        cv2.rectangle(img1, (20, 20 + i*60), (80, 70 + i*60), (0, 0, 0), -1)

    # Figura 2: 1 grande na esquerda, 4 na direita
    cv2.rectangle(img2, (20, 20), (80, 180), (0, 0, 0), -1)
    for i in range(4):
        cv2.rectangle(img2, (120, 20 + i*45), (180, 55 + i*45), (0, 0, 0), -1)

    # Figura 3: 1 em cima, 1 embaixo (simulando a Figura do enigma)
    # Vamos desenhar os 7 quadrados para bater com a sua lógica
    # (Para simplificar, vamos desenhar 7 quadrados espalhados)
    posicoes = [(20,20), (80,20), (20,80), (80,80), (20,140), (80,140), (50,80)]
    for (x, y) in posicoes:
        cv2.rectangle(img3, (x, y), (x+40, y+40), (0, 0, 0), -1)

    cv2.imwrite('figura1.png', img1)
    cv2.imwrite('figura2.png', img2)
    cv2.imwrite('figura3.png', img3)
    print("Imagens de teste criadas com espaço entre os quadrados!")

# ==========================================
# PARTE 2: Contando quadrados usando BORDAS (Melhorado)
# ==========================================
def contar_quadrados_preenchidos(caminho_imagem):
    img = cv2.imread(caminho_imagem, cv2.IMREAD_GRAYSCALE)

    # 1. Binarização
    _, thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)

    # 2. Encontra os contornos
    contornos, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # 3. Filtro de área
    quadrados_validos = [c for c in contornos if cv2.contourArea(c) > 100]

    return len(quadrados_validos)

# ==========================================
# PARTE 3: A SUA REGRA DE PROGRESSÃO
# ==========================================
def sua_logica_de_progressao(qtd_preenchidos):
    # Base: 3 = 18. Regra: cada +1 preenchido, +1 no valor.
    valor_final = 18 + (qtd_preenchidos - 3)
    return valor_final

# ==========================================
# PARTE 4: Executando tudo
# ==========================================
criar_imagens_teste()
print("\n--- Resolvendo o Enigma ---")

for i in range(1, 4):
    arquivo = f'figura{i}.png'
    qtd = contar_quadrados_preenchidos(arquivo)

    # Agora a sua lógica é aplicada ao resultado da contagem
    resultado = sua_logica_de_progressao(qtd)

    print(f"Figura {i}: Encontrou {qtd} quadrados preenchidos -> Resultado: {resultado}")
```
